```python
import pandas as pd
import time
import os
import ddddocr
from selenium import webdriver 
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from bs4 import BeautifulSoup
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
```

```python
driver = webdriver.Chrome()
wait = WebDriverWait(driver, 10)
driver.get('https://rucaptcha.com/demo/normal')
driver.switch_to.new_window('tab')
driver.get('https://www.prepostseo.com/ru/image-to-text')
handles = driver.window_handles
```

```python
# заполнение формы, которую нашли по XPATH. интересовать нас будет естественно контейнер input 
# ====================================================================================
# поиск и сохранение картинки на странице
# Находим капчу: //*[@id="capcher"]
driver.switch_to.window(handles[0])
os.makedirs('capcha_temp', exist_ok=True)
captcha = wait.until(
    EC.presence_of_element_located((By.CLASS_NAME, '_captchaImage_rrn3u_9'))
)
# Сохраняем скриншот только этого элемента
captcha.screenshot('capcha_temp/capcha.png')
# Распознование через стороннюю библиотеку ddddocr
# ocr = ddddocr.DdddOcr()
# with open('capcha_temp/capcha.png', 'rb') as file:
#      img_bytes = file.read()
# captcha_text = ocr.classification(img_bytes)
# print('Капча:', captcha_text)
# Распознавание капчи онлайн
# https://www.prepostseo.com/ru/image-to-text
```

```python
for inp in driver.find_elements(By.TAG_NAME, 'input'):
    print(
        'type=', inp.get_attribute('type'),
        '| id=', inp.get_attribute('id'),
        '| name=', inp.get_attribute('name'),
        '| class=', inp.get_attribute('class')
    )
```

```python
driver.switch_to.window(handles[1])
input_form = wait.until(EC.presence_of_element_located((By.ID, 'uploadFile')))     
input_form.send_keys(os.path.abspath(os.path.join('capcha_temp', 'capcha.png')))
hidden_button = wait.until(EC.element_to_be_clickable((By.CLASS_NAME, 'extract-text')))
hidden_button.click()
```

```python
get_text = wait.until(EC.presence_of_element_located((By.CLASS_NAME, 'result__text')))
text = get_text.text
```

```python
driver.switch_to.window(handles[0])
input_field = wait.until(EC.element_to_be_clickable((By.CLASS_NAME, '_inputInner_ws73z_12')))
input_field.clear()
input_field.send_keys(text)
check_button = wait.until(EC.element_to_be_clickable((By.CLASS_NAME, '_actionsItem_151cx_41')))
check_button.click()
```
