```python
from selenium import webdriver 
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import pandas as pd
import time
from bs4 import BeautifulSoup
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
```

```python
from selenium import webdriver 

driver = webdriver.Chrome()
driver.get("https://catalog.onliner.by/mobile")
```

```python
def get_all_links_on_page():
    elements = driver.find_elements(By.CLASS_NAME, value = 'catalog-form__link')

    ads_list = []
    for element in elements:
        ads_list.append(element.get_attribute('href'))
    
    return ads_list
```

```python
links = get_all_links_on_page()
```

```python
# Переход на следующую страницу через нажатие на кнопку
def click_next_page():
    clickable_element = WebDriverWait(driver, 10).until(
        EC.element_to_be_clickable((By.XPATH, '/html/body/div[2]/div/div/div/div/div/div/div/div/div[2]/div[1]/div/div/div[4]/div/div/div/div/div[3]/div[3]/a'))
    )

    clickable_element.click()
```

```python
click_next_page()
```

```python
clear_links = []

for item in links:
    if item == None or 'forum.' in item or 'tech.' in item:
        pass
    else:
        clear_links.append(item)

len(clear_links)
```

```python
very_clear_links = []

for item in clear_links:
    if 'prices' in item or 'reviews' in item or 'used' in item:
        pass
    else:
        very_clear_links.append(item)

len(very_clear_links)
```

```python
def clean_links(links: list):
    print(f'Start size of links list: {len(links)}')
    clear_links = []
    very_clear_links = []

    for item in links:
        if item == None or 'forum.' in item or 'tech.' in item:
            pass
        else:
            clear_links.append(item)
    print(f'Current size of links list: {len(clear_links)}')

    for item in clear_links:
        if 'prices' in item or 'reviews' in item or 'used' in item:
            pass
        else:
            very_clear_links.append(item)
    print(f'Finish size of links list: {len(very_clear_links)}')

    return very_clear_links
```

```python
links_list = []
try:
    for iter in range(0, 5, 1):
        links_list += get_all_links_on_page()
        click_next_page()
except:
    print("All done")
    clean_links_list = clean_links(links_list)
finally:
    print("All done")
    clean_links_list = clean_links(links_list)
```

```python
# Собираем все характеристики на объявления с выбранной страницы
def get_characteristic():
    character_dict = {}
    elements = driver.find_elements(By.CLASS_NAME, value='product-specs__table')
    for item in elements:
        tds = item.find_elements(By.TAG_NAME, value='td')
        for i in range(0, len(tds)-2, 2):
            character_dict.update({tds[i].text:tds[i+1].text})
    for i in character_dict:
        print(i)

get_characteristic()
```

```python
# Собираем все характеристики на объявления с выбранной страницы
def get_characterist():
    character_dict = {}
    elements = driver.find_elements(By.CSS_SELECTOR, value='.product-specs__table tr')
    for item in elements:
        tds = item.find_elements(By.TAG_NAME, value='td')
        if len(tds) == 2:
            character_dict.update({tds[0].text:tds[1].text})
    for k, v in character_dict.items():
        print(f'{k}: {v}')
        
get_characterist()
```

```python
# Собираем все характеристики на объявления с выбранной страницы
def get_characteristics():
    character_list = []
    character_list.append({'Название модели':driver.find_element(By.CLASS_NAME, value='catalog-masthead__title').text})
    elements = driver.find_elements(By.CSS_SELECTOR, value='.product-specs__table tr')
    for item in elements:
        tds = item.find_elements(By.TAG_NAME, value='td')
        if len(tds) == 2:
            if not tds[1].text:
                if tds[1].find_elements(By.CLASS_NAME, value="i-x"):
                    character_list.append({tds[0].text:'NO'})
                else:
                    character_list.append({tds[0].text:'Yes'})
                                          
                character_list.append({tds[0].text:tds[1].text})
            else:
                character_list.append({tds[0].text:tds[1].text})
    return character_list
```

```python
table = pd.DataFrame()
for l in clean_links_list[0:3]:
    driver.get(l)

    new_row = pd.DataFrame([get_characteristics()])
    table = pd.concat([table, new_row], ignore_index=True)
    


table
```
